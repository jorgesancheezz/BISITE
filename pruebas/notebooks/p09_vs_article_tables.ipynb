```python
import os, pandas as pd
from IPython.display import display, HTML

def show_csv(title, csvp, htmlp=None):
    if os.path.exists(csvp):
        df = pd.read_csv(csvp)
        display(HTML(f'<h2>{title}</h2>'))
        display(HTML(df.to_html(index=False, float_format=\"{:.4f}\".format)))
        print('Saved CSV:', csvp)
        if htmlp and os.path.exists(htmlp):
            print('Also saved HTML at', htmlp)
    else:
        print('Missing', csvp)

show_csv('P09 processed vs Article', 'compare_out_generated/p09_processed_vs_article_metrics.csv', 'compare_out_generated/p09_processed_vs_article_metrics.html')
show_csv('P09 / P10 vs Article', 'compare_out_generated/p09_p10_vs_article_metrics.csv', 'compare_out_generated/p09_p10_vs_article_metrics.html')
```
